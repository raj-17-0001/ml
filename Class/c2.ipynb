```python
def variance(*args):
    sum=0
    for i in args:
        sum+=(i-np.mean(args))**2
        calculated_variance=sum/(len(args))
    return calculated_variance
```

```python
def sample_variance(*args):
    sum=0
    for i in args:
        sum+=(i-np.mean(args))**2
        calculated_variance=sum/(len(args)-1)
    return calculated_variance
```
