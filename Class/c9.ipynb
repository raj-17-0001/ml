```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

from statsmodels.stats.outliers_influence import (variance_inflation_factor)

# Step 2: Create dataset
df = pd.DataFrame({
    "Area_sqft": [
        1000, 1500, 2000, 2500, 3000, 3500
    ],
    "Area_sqm": [
        92.9, 139.4, 185.8, 232.3, 278.7, 325.2
    ],
    "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2]
})

#step 3: Calculate VIF
X = df.copy()
vif_data = pd.DataFrame()
vif_data["feature"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print("\nResult: ")
vif_data.round(2)
```

```python
#creating OLS model
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Area_sqft": [
        1000, 1500, 2000, 2500, 3000, 3500
    ],
   "floor": [2, 3, 2, 4, 3, 2],
    "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2],
    "price": [120000, 130000, 140000, 150000, 160000, 170000]
})
```

```python
import statsmodels.api as sm
```

```python
x= df.iloc[:,:-1]
y= df.iloc[:,-1]
```

```python
x=sm.add_constant(x)
```

```python
model = sm.OLS(y, x).fit()
print(model.summary())
```
