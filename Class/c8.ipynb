```python

import numpy as np

x1 = np.array([2, 3, 4, 5, 6, 7])
x2 = np.array([12, 13, 14, 15, 16, 17])
x3 = np.array([3, 4, 5, 6, 8, 6])
y = np.array([17, 20, 23, 26, 29, 30])

m = len(x1)

w1 = 0
w2 = 0
w3 = 0
b = 0

learning_rate = 0.001
epochs = 1000

for i in range(epochs):

    y_pred = w1*x1 + w2*x2 + w3*x3 + b

    error = y - y_pred

    w1d = (2/m) * np.sum(error*(-x1))
    w2d = (2/m) * np.sum(error*(-x2))
    w3d = (2/m) * np.sum(error*(-x3))
    bd = (2/m) * np.sum(error*(-1))

    w1 = w1 - learning_rate*w1d
    w2 = w2 - learning_rate*w2d
    w3 = w3 - learning_rate*w3d
    b = b - learning_rate*bd

    if i % 100 == 0:
        print(f"Epoch: {i}, MSE: {np.mean(error**2)}")
        
y_pred = w1*x1 + w2*x2 +w3*x3 + b
error = y - y_pred

print("\nFinal Results")
print(f"w1: {w1}")
print(f"w2: {w2}")
print(f"w3: {w3}")
print(f"b: {b}")
print(f"Predictions: {y_pred}")
print(f"Final MSE: {np.mean(error**2)}")
```
