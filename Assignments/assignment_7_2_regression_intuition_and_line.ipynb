Question 1: OLS Mathematical Challenge
A data scientist collects the following dataset to predict students' exam marks based on study hours.

Tasks:
Calculate the mean of x and y, derive the OLS slope and intercept, and construct the regression equation from scratch.
Calculate the predicted marks and residuals for all six students.
Calculate SSE and MSE, then explain why minimizing squared residuals can produce a different result from minimizing absolute residuals.
Suppose the sixth student's actual marks change from 10 to 25. Recalculate the regression coefficients and analyze how this outlier affects the slope, intercept, and prediction accuracy.


```python
#task 1
import numpy as np
x = np.array([1, 2, 4, 5, 7, 8], dtype=float)
y = np.array([3, 4, 7, 8, 12, 10], dtype=float)

x_mean = np.mean(x)
y_mean = np.mean(y)

numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)

slope = numerator / denominator
intercept = y_mean - slope * x_mean

print("Mean of x:", round(x_mean, 4))
print("Mean of y:", round(y_mean, 4))
print("Numerator:", round(numerator, 4))
print("Denominator:", round(denominator, 4))
print("Slope:", round(slope, 4))
print("Intercept:", round(intercept, 4))

print(f"Regression Equation: y = {intercept:.2f} + {slope:.2f}x")
```

Question 2: Build a Regression Model from Scratch
You are given the following dataset:

import numpy as np
x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([15, 20, 28, 35, 43, 48], dtype=float)
Tasks:

Implement simple linear regression using only NumPy to calculate the slope and intercept, without using scikit-learn.
Predict the target values for x = 5, x = 9, and x = 16, identifying which predictions are interpolation and extrapolation.
Calculate residuals, SSE, MSE, and RMSE on the original dataset.

```python
import numpy as np
x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([15, 20, 28, 35, 43, 48], dtype=float)
x_mean = np.mean(x)
y_mean = np.mean(y)

slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
intercept = y_mean - slope * x_mean

print("Slope:", round(slope, 4))
print("Intercept:", round(intercept, 4))

new_x = np.array([5, 9, 16], dtype=float)
new_predictions = intercept + slope * new_x

print("\nNew Predictions:")
for value, prediction in zip(new_x, new_predictions):
    print(f"x = {value:.0f}, predicted y = {prediction:.4f}")

y_pred = intercept + slope * x
residuals = y - y_pred
SSE = np.sum(residuals ** 2)
MSE = SSE / len(y)
RMSE = np.sqrt(MSE)

print("\nPredictions:", np.round(y_pred, 4))
print("Residuals:", np.round(residuals, 4))
print("SSE:", round(SSE, 4))
print("MSE:", round(MSE, 4))
print("RMSE:", round(RMSE, 4))
```

Question 3: Regression Model Selection and Extrapolation Risk
A manufacturing company predicts machine fuel consumption using operating hours. The training dataset contains operating hours between 10 and 100.

The fitted model is:

Fuel_Consumption = 25 + 1.8 * Operating_Hours
A manager wants predictions for machines operating for 40, 85, 120, and 200 hours.

Tasks:

Calculate all four predicted fuel-consumption values and classify each prediction as interpolation or extrapolation.
Explain why predictions at 120 and 200 hours may be unreliable even if the model has a high training (R^2).

```python
intercept = 25
slope = 1.8

hours = [40, 85, 120, 200]

for h in hours:
    fuel = intercept + slope * h

    if 10 <= h <= 100:
        prediction_type = "Interpolation"
    else:
        prediction_type = "Extrapolation"

    print(
        f"Operating Hours: {h}, "
        f"Predicted Fuel: {fuel:.2f}, "
        f"Type: {prediction_type}"
    )
```
