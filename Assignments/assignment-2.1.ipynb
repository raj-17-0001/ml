Q1. Employee Salary Analysis A company records the monthly salaries of its employees, in thousands of rupees.

(a) Write Python functions to calculate the mean, median, and all modes without using statistics.mean(), statistics.median(), or statistics.mode(). Display the results for the original dataset.

```python
salaries = [    18, 20, 22, 22, 24,    25, 25, 26, 28, 30,    32, 35, 38, 42, 150]
```

```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    data = sorted(data)
    n = len(data)
    
    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_modes(data):
    counts = {}
    
    for x in data:
        counts[x] = counts.get(x, 0) + 1
    
    highest = max(counts.values())
    
    if highest == 1:
        return []
    
    return [x for x in counts if counts[x] == highest]

salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]
print("Mean:", calculate_mean(salaries))
print("Median:", calculate_median(salaries))
print("Modes:", calculate_modes(salaries))
```

(b) Remove the salary of 150 and recalculate all three measures. Calculate the percentage change in the mean and median, then justify which measure better represents a typical employee's salary.

```python
clean_salaries = [x for x in salaries if x != 150]

old_mean = calculate_mean(salaries)
old_median = calculate_median(salaries)

new_mean = calculate_mean(clean_salaries)
new_median = calculate_median(clean_salaries)

mean_change = (new_mean - old_mean) / old_mean * 100
median_change = (new_median - old_median) / old_median * 100

print("New Mean:", round(new_mean, 2))
print("New Median:", new_median)
print("Modes:", calculate_modes(clean_salaries))
print("Mean Percentage Change:", round(mean_change, 2), "%")
print("Median Percentage Change:", round(median_change, 2), "%")
```

Q2. Customer Spending Analysis Using the Tips Dataset Use Seaborn's built-in tips dataset.

(a) Calculate the mean, median, and all modes of total_bill and tip. Then divide the dataset into Lunch and Dinner groups and calculate the mean and median bill for each group.

```python
import seaborn as sns

tips = sns.load_dataset("tips")

for column in ["total_bill", "tip"]:
    values = tips[column].tolist()

    print("\nColumn:", column)
    print("Mean:", round(calculate_mean(values), 3))
    print("Median:", calculate_median(values))
    print("All Modes:", calculate_modes(values))

for meal in ["Lunch", "Dinner"]:
    bills = tips.loc[
        tips["time"] == meal, "total_bill"
    ].tolist()

    print("\nMeal:", meal)
    print("Number of bills:", len(bills))
    print("Mean Bill:", round(calculate_mean(bills), 3))
    print("Median Bill:", calculate_median(bills))
```

(b) Identify the three largest bills, temporarily exclude them, and recalculate the overall mean and median. Use Python to compare the results before and after exclusion, and explain which measure better represents typical customer spending.

```python
largest_indexes = tips["total_bill"].sort_values(
    ascending=False
).head(3).index

tips_without_top3 = tips.drop(index=largest_indexes)

before_bills = tips["total_bill"].tolist()
after_bills = tips_without_top3["total_bill"].tolist()

print("Three largest bills:")
print(tips.loc[largest_indexes, "total_bill"].sort_values(
    ascending=False
).tolist())

print("\nBefore exclusion:")
print("Mean:", round(calculate_mean(before_bills), 3))
print("Median:", calculate_median(before_bills))

print("\nAfter exclusion:")
print("Mean:", round(calculate_mean(after_bills), 3))
print("Median:", calculate_median(after_bills))
```

Q3. Build a Central Tendency Calculator From Scratch Use the following datasets:

(a) Build reusable Python functions to calculate the mean, median, and mode without using built-in statistical functions. Your program must handle unsorted data, odd and even dataset lengths, multiple modes, and datasets with no repeated values.

(b) Run your calculator on all five datasets and generate a formatted report. Identify which dataset is most affected by an extreme value and demonstrate numerically why its mean and median behave differently.

```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    values = sorted(data)
    n = len(values)

    if n % 2 == 1:
        return values[n // 2]
    else:
        return (values[n // 2 - 1] + values[n // 2]) / 2

def calculate_modes(data):
    counts = {}

    for value in data:
        counts[value] = counts.get(value, 0) + 1

    highest = max(counts.values())

    if highest == 1:
        return []

    modes = [
        value for value, count in counts.items()
        if count == highest
    ]

    return sorted(modes)


datasets = {
    "data_1": [12, 15, 15, 18, 20, 22, 25],
    "data_2": [5, 5, 10, 10, 15, 20],
    "data_3": [7, 14, 21, 28, 35],
    "data_4": [100, 100, 100, 100],
    "data_5": [1, 2, 3, 4, 5, 100]
}

print(f"{'Dataset':<10} {'Mean':>10} {'Median':>10} {'Modes':>12}")
print("-" * 45)

for name, data in datasets.items():
    mean = calculate_mean(data)
    median = calculate_median(data)
    modes = calculate_modes(data)

    if modes == []:
        modes = "No mode"

    print(f"{name:<10} {mean:>10.2f} {median:>10.2f} {str(modes):>12}")


data_5 = datasets["data_5"]
clean_data = [x for x in data_5 if x != 100]

print("\nData_5 Before Removing 100")
print("Mean:", round(calculate_mean(data_5), 2))
print("Median:", calculate_median(data_5))

print("\nData_5 After Removing 100")
print("Mean:", round(calculate_mean(clean_data), 2))
print("Median:", calculate_median(clean_data))
```

Q4. Statistical Transformations and Reasoning An online store records the number of products purchased per order.

(a) Calculate the mean, median, and all modes. Remove the largest observation and recalculate the measures. Use the results to explain how an extreme order affects each measure.

```python
orders = [ 1, 1, 1, 2, 2, 2, 2, 3, 3, 4, 4, 5, 5, 6, 8, 10, 15, 20, 25, 80] 
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    values = sorted(data)
    n = len(values)

    if n % 2 == 1:
        return values[n // 2]
    else:
        return (values[n // 2 - 1] + values[n // 2]) / 2

def calculate_modes(data):
    counts = {}

    for value in data:
        counts[value] = counts.get(value, 0) + 1

    highest = max(counts.values())

    return sorted([
        value for value, count in counts.items()
        if count == highest
    ])


print("Original Data")
print("Mean:", calculate_mean(orders))
print("Median:", calculate_median(orders))
print("Modes:", calculate_modes(orders))


new_orders = orders[:-1]

print("\nAfter Removing Largest Observation")
print("Mean:", round(calculate_mean(new_orders), 2))
print("Median:", calculate_median(new_orders))
print("Modes:", calculate_modes(new_orders))
```

(b) Create two transformed datasets: add 20 to every observation in the first transformation and multiply every observation by 3 in the second. Calculate the mean, median, and mode for each dataset, then use Python to verify how these transformations affect the original measures.

```python
from math import isclose

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_modes(data):
    counts = {}

    for x in data:
        counts[x] = counts.get(x, 0) + 1

    highest = max(counts.values())
    return sorted([x for x in counts if counts[x] == highest])


# Transformations
add_20 = [x + 20 for x in orders]
multiply_3 = [x * 3 for x in orders]

mean = calculate_mean(orders)
median = calculate_median(orders)
modes = calculate_modes(orders)

print("Original Data:", mean, median, modes)

print("After Adding 20:")
print("Mean:", calculate_mean(add_20))
print("Median:", calculate_median(add_20))
print("Mode:", calculate_modes(add_20))

print("\nAfter Multiplying by 3:")
print("Mean:", calculate_mean(multiply_3))
print("Median:", calculate_median(multiply_3))
print("Mode:", calculate_modes(multiply_3))

# Verify results
assert isclose(calculate_mean(add_20), mean + 20)
assert calculate_median(add_20) == median + 20
assert calculate_modes(add_20) == [x + 20 for x in modes]

assert isclose(calculate_mean(multiply_3), mean * 3)
assert calculate_median(multiply_3) == median * 3
assert calculate_modes(multiply_3) == [x * 3 for x in modes]
```

Q5. Missing Values and Outlier Investigation A data analyst receives the following daily sales figures, in thousands of rupees.

(a) Convert the data into a Pandas Series, identify missing observations, and calculate the percentage of missing values. Excluding missing values, calculate the mean, median, and all modes. Then use the IQR method to identify potential outliers.

```python
sales = [ 12, 15, 14, 16, None, 15, 18, 14, 17, 16, None, 19, 15, 120, 14]

import pandas as pd
s = pd.Series(sales)

print("Missing Values:")
print(s[s.isnull()])

missing = s.isnull().sum()
percentage = missing / len(s) * 100

print("Missing Count:", missing)
print("Missing Percentage:", round(percentage, 2), "%")

data = s.dropna().tolist()

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    values = sorted(data)
    n = len(values)

    if n % 2 == 1:
        return values[n // 2]
    return (values[n // 2 - 1] + values[n // 2]) / 2

def calculate_modes(data):
    counts = {}

    for x in data:
        counts[x] = counts.get(x, 0) + 1

    highest = max(counts.values())

    if highest == 1:
        return []

    return sorted([x for x in counts if counts[x] == highest])

print("\nCentral Tendency:")
print("Mean:", round(calculate_mean(data), 2))
print("Median:", calculate_median(data))
print("Modes:", calculate_modes(data))

q1 = s.dropna().quantile(0.25)
q3 = s.dropna().quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outliers = [x for x in data if x < lower_limit or x > upper_limit]

print("\nOutlier Detection:")
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Potential Outliers:", outliers)
```

(b) Recalculate the mean, median, and mode after excluding the suspected outlier of 120. Compare the results, calculate the percentage changes in the mean and median, and write a Python-supported recommendation on whether the outlier should be removed, retained, or investigated further.

```python
old_mean = calculate_mean(data)
old_median = calculate_median(data)
old_modes = calculate_modes(data)

# Remove the suspected outlier
clean_data = [x for x in data if x != 120]

new_mean = calculate_mean(clean_data)
new_median = calculate_median(clean_data)
new_modes = calculate_modes(clean_data)

# Percentage changes
mean_change = (new_mean - old_mean) / old_mean * 100
median_change = (new_median - old_median) / old_median * 100

print("Before Removing 120")
print("Mean:", round(old_mean, 2))
print("Median:", old_median)
print("Modes:", old_modes)

print("\nAfter Removing 120")
print("Mean:", round(new_mean, 2))
print("Median:", new_median)
print("Modes:", new_modes)

print("\nPercentage Change:")
print("Mean:", round(mean_change, 2), "%")
print("Median:", round(median_change, 2), "%")

print("\nRecommendation: Investigate the value 120 before removing it.")
```
