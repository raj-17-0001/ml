Question 1: Multiple Linear Regression and Coefficient Interpretation
A real-estate company wants to predict house prices using area, number of bedrooms, house age, and location score.

The trained model is:

Price = 12 + 0.045 * Area + 6 * Bedrooms - 1.2 * Age + 8 * Location_Score
Price is measured in lakhs of rupees, area in square feet, and age in years.

Tasks:

Calculate the predicted price for each house using the given equation.
Interpret all four feature coefficients while holding the other features constant.
Suppose the area coefficient changes from 0.045 to 0.060. Calculate the change in predicted price for House B, assuming all other coefficients remain unchanged.
Explain why a larger coefficient does not necessarily mean a feature is more important when features use different measurement scales.

```python
Price_A = 12+0.045*1200+6*2-1.2*10+8
Price_B = 12+0.045*1800+6*3-1.2*5+8*4
Price_C = 12+0.045*2400+6*4-1.2*8+8*5
Price_D = 12+0.045*1500+6*3-1.2*12+8*2
print("Hourse A Price:",Price_A,"Lakhs")
print("Hourse B Price:",Price_B,"Lakhs")
print("Hourse C Price:",Price_C,"Lakhs")
print("Hourse D Price:",Price_D,"Lakhs")
```
