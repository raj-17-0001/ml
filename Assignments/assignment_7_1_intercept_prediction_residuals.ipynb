Question 2: Integrated Regression Model Challenge
An automobile company builds a multiple linear regression model to predict used-car prices using vehicle age, mileage, and engine capacity. After training, the company observes that the model has a high \(R^2\) score but still makes large errors for expensive cars.

Tasks:

Explain how coefficients and the intercept determine the model's predictions, and identify a limitation of interpreting coefficients when input features are correlated.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

data = {"Age": [1, 2, 3, 4, 5, 6, 7, 8, 2, 4, 6, 9],
       "Mileage": [10000, 20000, 30000, 40000, 50000, 60000,70000, 80000, 18000, 42000, 65000, 95000],
       "Engine_CC": [1200, 1200, 1500, 1500, 1800, 1800,2000, 2000, 1500, 1800, 2000, 2200],
       "Price_Lakhs": [8, 7.5, 7, 6.5, 6, 5.5,5, 4.5, 9, 7, 5.8, 4]}

df = pd.DataFrame(data)
X = df[["Age", "Mileage", "Engine_CC"]]
y = df["Price_Lakhs"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

print("Intercept:", round(model.intercept_, 3))
for feature, coefficient in zip(X.columns, model.coef_):
    print(f"{feature} coefficient: {coefficient:.6f}")

y_pred = model.predict(X_test)

print("\nR2 Score:", round(r2_score(y_test, y_pred), 3))
print("MAE:", round(mean_absolute_error(y_test, y_pred), 3), "lakhs")
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred)), 3), "lakhs")

results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

print("\nActual vs Predicted:")
print(results.round(2))

print("\nFeature Correlation:")
print(df[["Age", "Mileage", "Engine_CC"]].corr().round(2))
```
