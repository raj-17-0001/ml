Question 1: Comparing Mean, Median, and Mode Imputation
A retail company has collected customer information, but several values are missing. Some numerical columns contain extreme values, while a categorical column contains repeated categories

A. Identify the missing values in every column. Apply mean imputation and median imputation separately to the numerical columns, and mode imputation to the categorical columns. Create separate completed datasets and compare the imputed values, especially for Age and Monthly_Income, where extreme values may affect the mean.

B. Calculate the mean, median, and standard deviation of the numerical columns before and after each imputation strategy. Explain how the choice of strategy changes the statistical characteristics of the data and justify which strategy is more appropriate for each column.

```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 35, 100, np.nan, 28],
    "Monthly_Income": [20000, 25000, 30000, np.nan,
                       35000, 500000, 28000, np.nan],
    "City": ["Delhi", "Mumbai", "Delhi", np.nan,
             "Pune", "Delhi", np.nan, "Mumbai"],
    "Purchase_Category": ["Electronics", "Clothing", np.nan,
                          "Electronics", "Clothing", "Electronics",
                          "Clothing", np.nan]
})

print("ORIGINAL DATA:")
print(df)
print("\nMISSING VALUES IN EACH COLUMN:")
print(df.isnull().sum())

numeric_cols = ["Age", "Monthly_Income"]
categorical_cols = ["City", "Purchase_Category"]

print("\nSTATISTICS BEFORE IMPUTATION:")
print(df[numeric_cols].agg(["mean", "median", "std"]).round(2))

mean_df = df.copy()
median_df = df.copy()
mode_df = df.copy()

mean_imputer = SimpleImputer(strategy="mean")
mean_df[numeric_cols] = mean_imputer.fit_transform(
    df[numeric_cols]
)

median_imputer = SimpleImputer(strategy="median")
median_df[numeric_cols] = median_imputer.fit_transform(
    df[numeric_cols]
)

mode_imputer = SimpleImputer(strategy="most_frequent")

for data in [mean_df, median_df, mode_df]:
    data[categorical_cols] = mode_imputer.fit_transform(
        data[categorical_cols]
    )


print("\nMEAN IMPUTATION DATASET:")
print(mean_df)

print("\nMEDIAN IMPUTATION DATASET:")
print(median_df)

print("\nMODE IMPUTATION DATASET:")
print(mode_df)
before_stats = df[numeric_cols].agg(["mean", "median", "std"])

mean_stats = mean_df[numeric_cols].agg(
    ["mean", "median", "std"]
)

median_stats = median_df[numeric_cols].agg(
    ["mean", "median", "std"]
)

mode_stats = mode_df[numeric_cols].agg(
    ["mean", "median", "std"]
)

print("\nBEFORE IMPUTATION:")
print(before_stats.round(2))

print("\nAFTER MEAN IMPUTATION:")
print(mean_stats.round(2))

print("\nAFTER MEDIAN IMPUTATION:")
print(median_stats.round(2))

print("\nAFTER MODE IMPUTATION:")
print(mode_stats.round(2))
```

Question 2: Implementing SimpleImputer with Different Strategies

A healthcare analytics team wants to prepare a dataset containing numerical measurements and categorical patient information.

A. Use SimpleImputer to create a preprocessing workflow that applies mean to Patient_Age, median to Blood_Sugar, and most_frequent to both categorical columns. Display the completed DataFrame with its original column names and verify that no missing values remain.

B. Create a second workflow that uses strategy="constant" to replace missing categorical values with "Unknown" and missing numerical values with 0. Compare the results of both workflows and discuss why replacing missing blood sugar measurements with zero may lead to misleading conclusions

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
    "Patient_Age": [25, 32, np.nan, 45, 52, np.nan, 38, 60],
    "Blood_Sugar": [90, 110, 105, np.nan, 250, 130, np.nan, 115],
    "Blood_Group": ["A", "B", "O", np.nan, "A", "B", np.nan, "A"],
    "Treatment": ["Basic", np.nan, "Advanced", "Basic",
                  "Advanced", "Basic", np.nan, "Basic"]
})

print("Original Data:")
print(df)

mean_cols = ["Patient_Age"]
median_cols = ["Blood_Sugar"]
categorical_cols = ["Blood_Group", "Treatment"]

mean_imputer = SimpleImputer(strategy="mean")
median_imputer = SimpleImputer(strategy="median")
mode_imputer = SimpleImputer(strategy="most_frequent")

preprocessor = ColumnTransformer(
    transformers=[
        ("age", mean_imputer, mean_cols),
        ("sugar", median_imputer, median_cols),
        ("category", mode_imputer, categorical_cols)
    ],
    remainder="passthrough",
    verbose_feature_names_out=False
)

result = preprocessor.fit_transform(df)

completed_df = pd.DataFrame(
    result,
    columns=preprocessor.get_feature_names_out(),
    index=df.index
)

completed_df = completed_df[df.columns]

completed_df["Patient_Age"] = pd.to_numeric(
    completed_df["Patient_Age"]
)
completed_df["Blood_Sugar"] = pd.to_numeric(
    completed_df["Blood_Sugar"]
)

print("\nCompleted DataFrame (Part A):")
print(completed_df)

print("\nMissing Values After Imputation:")
print(completed_df.isnull().sum())

print("\nAre there any missing values?")
print(completed_df.isnull().values.any())
```

```python

# part b 
numeric_cols = ["Patient_Age", "Blood_Sugar"]
categorical_cols = ["Blood_Group", "Treatment"]

numeric_constant_imputer = SimpleImputer(
    strategy="constant",
    fill_value=0
)

categorical_constant_imputer = SimpleImputer(
    strategy="constant",
    fill_value="Unknown"
)

constant_preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_constant_imputer, numeric_cols),
        ("categorical", categorical_constant_imputer, categorical_cols)
    ],
    verbose_feature_names_out=False
)

constant_result = constant_preprocessor.fit_transform(df)

constant_df = pd.DataFrame(
    constant_result,
    columns=constant_preprocessor.get_feature_names_out(),
    index=df.index
)

constant_df = constant_df[df.columns]
constant_df[numeric_cols] = constant_df[numeric_cols].apply(
    pd.to_numeric
)

print("Constant Imputation DataFrame (Part B):")
print(constant_df)

print("\nMissing Values:")
print(constant_df.isnull().sum())

print("\nAre there any missing values?")
print(constant_df.isnull().values.any())

print("\nPart A: Statistical Imputation")
print(completed_df)

print("\nPart B: Constant Imputation")
print(constant_df)
```

Question 3: Investigating the Impact of Outliers on Imputation
An online marketplace has recorded customer spending and order counts. A few customers have unusually high spending values, and several records are incomplete.

A. Impute missing numerical values using both mean and median strategies with SimpleImputer. Compare the imputed values and calculate the mean and standard deviation of Annual_Spending after each method. Explain the effect of the unusually large spending value on the mean.

B. Build a reusable preprocessing pipeline using ColumnTransformer and SimpleImputer that applies median imputation to numerical features and most-frequent imputation to categorical features. Exclude Customer_ID from the input features, then verify that the transformed dataset contains no missing values and retains the expected feature columns.

```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df = pd.DataFrame({
    "Customer_ID": range(1, 11),
    "Annual_Spending": [12000, 15000, np.nan, 18000, 22000,
                        25000, 30000, np.nan, 900000, 20000],
    "Order_Count": [4, 5, 6, np.nan, 8, 7, 9, 10, 50, np.nan],
    "Membership": ["Silver", "Gold", "Silver", "Gold", np.nan,
                   "Silver", "Gold", np.nan, "Platinum", "Silver"]
})

print("Original DataFrame:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())


numeric_cols = ["Annual_Spending", "Order_Count"]

mean_imputer = SimpleImputer(strategy="mean")
mean_data = mean_imputer.fit_transform(df[numeric_cols])

mean_df = pd.DataFrame(
    mean_data,
    columns=numeric_cols,
    index=df.index
)

median_imputer = SimpleImputer(strategy="median")
median_data = median_imputer.fit_transform(df[numeric_cols])

median_df = pd.DataFrame(
    median_data,
    columns=numeric_cols,
    index=df.index
)

print("\nMean Imputation Results:")
print(mean_df)

print("\nMedian Imputation Results:")
print(median_df)

comparison = pd.DataFrame({
    "Mean_Imputation": mean_df["Annual_Spending"].agg(
        ["mean", "std"]
    ),
    "Median_Imputation": median_df["Annual_Spending"].agg(
        ["mean", "std"]
    )
})

print("\nAnnual Spending Statistics Comparison:")
print(comparison.round(2))


X = df.drop(columns=["Customer_ID"])

numeric_features = ["Annual_Spending", "Order_Count"]
categorical_features = ["Membership"]

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_features),
        ("cat", categorical_pipeline, categorical_features)
    ],
    remainder="drop",
    verbose_feature_names_out=False
)

transformed_data = preprocessor.fit_transform(X)
processed_df = pd.DataFrame(
    transformed_data,
    columns=preprocessor.get_feature_names_out(),
    index=df.index
)

processed_df[numeric_features] = processed_df[numeric_features].apply(
    pd.to_numeric
)

print("\nPart B: Processed DataFrame:")
print(processed_df)

print("\nExpected Feature Columns:")
print(processed_df.columns.tolist())

print("\nMissing Values After Preprocessing:")
print(processed_df.isnull().sum())

print("\nAny Missing Values Remaining?")
print(processed_df.isnull().values.any())

print("\nIs Customer_ID Excluded?")
print("Customer_ID" not in processed_df.columns)
```
