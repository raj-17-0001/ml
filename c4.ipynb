```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.preprocessing import PowerTransformer
```

```python
#1. Box-Cox Transformation: Positive data only
arr = np.array([[1], [1], [3], [5], [10], [50]])
boxcox = PowerTransformer(method='box-cox')
arr_boxcox = boxcox.fit_transform(arr)
print("Original Data:\n", arr.ravel())
print("Transformed Data:\n", np.round(arr_boxcox.ravel(),3))
print("Lambda Parameter:\n", boxcox.lambdas_)
```

```python
#2. Yeo-Johnson Transformation: Works with positive and negative data

yeo_johnson = PowerTransformer(method='yeo-johnson')
arr_transformed = yeo_johnson.fit_transform(arr)
print("Original Data:\n", arr.ravel())
print("Transformed Data:\n", np.round(arr_transformed.ravel(), 3))
print("Lambda Parameter:\n", yeo_johnson.lambdas_)
```

```python
from sklearn.impute import SimpleImputer
```

```python
#filling missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona', 'George', 'Hannah'],
    
    # --- Categorical Columns (with nulls) ---
    'Department': ['IT', 'HR', np.nan, 'Marketing', 'IT', np.nan, 'Finance', 'HR'],
    'City': ['New York', np.nan, 'London', 'Paris', 'Tokyo', 'London', np.nan, 'Tokyo'],
    'Education': ['Bachelors', 'Masters', 'Bachelors', None, 'PhD', 'Masters', 'Bachelors', None],
    
    # --- Numerical Columns (with nulls) ---
    'Age': [25, np.nan, 30, 35, np.nan, 28, 42, 31],
    'Salary': [50000, 62000, np.nan, 58000, 95000, np.nan, 80000, 67000],
    'Experience_Years': [2, 5, 4, np.nan, 12, 3, np.nan, 6]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print("--- DataFrame ---")
print(df)
```

```python
imputer = SimpleImputer(strategy='mean')
df['New_Salary'] = imputer.fit_transform(df[['Salary']]).flatten()
```

```python
df
```

```python
imputer = SimpleImputer(strategy='median')
df['New_Experience_Years'] = imputer.fit_transform(df[['Experience_Years']]).flatten()
```

```python
df
```

```python
imputer = SimpleImputer(strategy='most_frequent')
df['New_Department'] = imputer.fit_transform(df[['Department']]).flatten()
```

```python
imputer = SimpleImputer(strategy='most_frequent')
df['City'] = imputer.fit_transform(df[['City']]).flatten()
```
